# Data Visualization

**# Zone Wise Gender Count**

```python
C1 = ["#00695C", "#59E4D4"]

plt.figure(figsize=(15,6))

A1 = sns.countplot(x="Zone",data= df,hue="Gender",palette=C1)

for i in A1.containers:
    A1.bar_label(i)

plt.title("Zone vs Gender Wise Count",color="#094E46",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

Conclusion : The bar chart shows that the Central zone has the highest count of individuals for both genders, while the Eastern zone has the lowest.

**# Age Group Wise Marital Status Count**

```python
C2 = ["#4B0082", "#A368E6"]

plt.figure(figsize=(15,6))

A2 = sns.countplot(x="Age Group",data= df,hue="Marital_Status",palette=C2,edgecolor="black")

for i in A2.containers:
    A2.bar_label(i)

plt.title("Age Group Wise Marital Status Count",color="#4B0082",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

Conclusion : The 26-35 age group represents the largest segment of the population by far, while the 0-17 age group represents the smallest. Across all age demographics, single individuals are consistently more common than married individuals.

**# Sector Wise Amount in Sorted format**

```python
linechart =df.groupby("Sector",as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
```

```python


plt.figure(figsize=(15,6))

sns.lineplot(x="Sector",y="Amount",data=linechart,marker="o",color="#0D3E0F")

plt.xticks(rotation=45)

plt.title("Sector Wise Amount in sorted format",color="#0D3E0F",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")

plt.ylabel("")

plt.grid()

plt.show()

```

Conclusion : The IT Sector has the highest total amount, while Agriculture has the lowest. The chart shows a steady upward trend in amounts as you move from primary production fields (like Textile and Construction) up to service and technology-driven industries (like Banking and Healthcare).

**# State Wise Total Orders**

```python
barchart = df.groupby("State",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False)
```

```python
plt.figure(figsize=(15,6))

A3= sns.barplot(x="State",y="Orders",data=barchart,palette="Set1")

for i in A3.containers:
    A3.bar_label(i)

plt.xticks(rotation=45)

plt.title("State Wise Total Orders",color="#771414",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

Conclusion : Uttar Pradesh has the highest number of total orders, while Telangana has the lowest.The data shows a clear downward pattern,where the top four states generate significantly more order volume than the remaining regions.

**# Zone Wise Total Users**

```python
plt.figure(figsize=(15,6))

C4 = ["#873D15", "#BB6118", "#F79C41", "#573809", "#B79055", "#FFD180"]

piechart = df.groupby("Zone",as_index=False)["User_ID"].count()

plt.pie(piechart["User_ID"],labels=piechart["Zone"],autopct="%.f%%",colors=C4)

plt.title("Zone Wise Total Users",color="#4A290D",fontweight="bold", fontsize=16, family="georgia")


plt.show()
```

Conclusion : The Central zone is largest market, bringing in more than one-third of all users at 38%. The Eastern zone is the smallest market, contributing only 7% of the total user base.


**# Age Group Wise Gender Count**

```python
plt.figure(figsize=(15,5))

c1=["#491655FF","#D6D257"]
A1=sns.countplot(x="Age Group",data=df,hue="Gender",palette=c1,edgecolor="black")

for y in A1.containers:
    A1.bar_label(y)


plt.title("Age Group Wise Gender Count",color="#491655FF",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")
plt.ylabel("")
plt.show()
```

**# Age Group Wise Total Orders**

```python
Data1 =df.groupby("Age Group",as_index=False)["Orders"].sum()

plt.figure(figsize=(15,5))

sns.lineplot(x="Age Group",y="Orders",data=Data1,marker="o",color="Black")

plt.xlabel("")
plt.ylabel("")
plt.title("Age Group Wise Total Orders",color="#200925FF",fontweight="bold", fontsize=16, family="georgia")
plt.show()
```

Conclusion : Orders spike dramatically and peak in the 26-35 age group,showing they are your most active shoppers. Buying activity is very low among the youngest (0-17) and oldest (55+) demographics.

**# Zone Wise Total Orders**

```python
data2=df.groupby("Zone",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False)

plt.figure(figsize=(15,6))
c4=["#1B6B5C"]
A3=sns.barplot(x="Zone",y="Orders",data=data2,errorbar=None,palette=c4,edgecolor="black")

for i in A3.containers:
    A3.bar_label(i)

plt.title("Zone Wise Total Orders",color="#1A7B69",fontweight="bold", fontsize=16, family="georgia")
plt.xlabel("")
plt.ylabel("")
plt.show()
```

Conclusion : the Central zone generates the highest volume of total orders. Order numbers drop steadily across the other regions, leaving the Eastern zone with the lowest sales volume.

**# Top 10 Product_Category**

```python
C1 = [
    "#0D47A1", "#1565C0", "#1976D2", "#1E88E5", "#42A5F5",
    "#64B5F6", "#90CAF9", "#A9D4FA", "#C5E1F9", "#DDEEFF"
]

plt.figure(figsize=(15,6))

data3 = df.groupby("Product_Category",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
A3= sns.barplot(x="Product_Category",y="Orders",data=data3,palette=C1)

for i in A3.containers:
    A3.bar_label(i)

plt.xticks(rotation=45)

plt.title("Top 10 Product_Category Wise Orders",color="#0F35A6",fontweight="bold", fontsize=16, family="georgia")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

Conclusion : Food, Clothing, and Electronics are your absolute best-selling product categories, holding the top three spots. The remaining seven categories see a sharp drop in order volume, showing much lower consumer demand.

**# Top 5 State According To Users**

```python
data4=df.groupby("State",as_index=False)["User_ID"].count().sort_values(by="User_ID",ascending=False).head(5)

plt.figure(figsize=(10,6))
c4=["#580D52"]
A3=sns.barplot(x="State",y="User_ID",data=data4,errorbar=None,palette=c4,edgecolor="black")

for i in A3.containers:
    A3.bar_label(i)

plt.title("State Wise Total Users",color="#5B1957",fontweight="bold", fontsize=16, family="georgia")
plt.xlabel("")
plt.ylabel("")
plt.show()
```
