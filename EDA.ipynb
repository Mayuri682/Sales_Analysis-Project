# Heatmap

**--> Gender & State Wise Orders**

```python
Pivot_table_1 = df.pivot_table(
values="Orders",
columns="Gender",
index="State",
aggfunc="sum")
```

```python
sns.heatmap(Pivot_table_1,annot=True,fmt="0.0f",cmap="YlGnBu")
```

**--> Zone & State Wise Orders**

```python
Pivot_table_3 = df.pivot_table(
values="Orders",
columns="Zone",
index="State",
aggfunc="sum")
```

```python
sns.heatmap(Pivot_table_3,annot=True,fmt="0.0f",cmap="YlGnBu")
```

**--> Marital Status & Sector Wise Orders**

```python
Pivot_table_4 = df.pivot_table(
values="Orders",
columns="Marital_Status",
index="Sector",
aggfunc="sum")
```

```python
sns.heatmap(Pivot_table_4,annot=True,fmt="0.0f")
```

**--> Product category & Zone Wise Orders**

```python
Pivot_table_5 = df.pivot_table(
values="Orders",
columns="Zone",
index="Product_Category",
aggfunc="sum")
```

```python
sns.heatmap(Pivot_table_5,annot=True,fmt="0.0f")
```
